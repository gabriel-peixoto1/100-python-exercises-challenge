Exercise 27: Function with Parameters
Objective: Learn how to pass parameters to a function.
Task:
• Define a function welcome(name) that prints a personalized greeting.
• Call the function with different names.
Expected Output:
Hello, Sara! Welcome to Lkhibra Academy!
Hello, Youssef! Welcome to Lkhibra Academy!

```python
def welcome(name):
    print(f"Hello, {name}! Welcome to Lkhibra Academy!\n")

a = "Gabriel"
b = "John"
c = "Jennie"
d = "Yuri"

welcome(a)
welcome(b)
welcome(c)
welcome(d)
```

Exercise 28: Function Returning a Value
Objective: Learn how a function can return a value.
Task:
• Define a function square(num) that returns the square of a number.
• Print the result.
Expected Output:
The square of 5 is 25.

```python
def square(num):
    return pow(num,2);


x = int(input("Enter a real number: "))
print(f"The square of {x} is {square(x)}.\n")
```

Exercise 29: Function with Default Parameter
Objective: Learn how to use default parameters in functions.
Task:
• Define a function greet(name="Student") that greets a user.
• Call it with and without a parameter.
Expected Output:
Hello, Student!
Hello, Amine!

```python
def greet(name = "Student"):
    print(f"Hello, {name}!\n")

greet()
greet("Anna")
```

Exercise 30: Function Using Global Variable
Objective: Learn how functions interact with global variables.
Task:
• Declare a global variable academy = "Lkhibra".
• Define a function that prints the value of the global variable.
Expected Output:
The best academy is Lkhibra!

```python
def value(txt):
    print(f"The best academy is {txt}!\n")

academy = "Lkhibra"
value(academy)
```
