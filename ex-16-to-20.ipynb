Exercise 16: Check if a Number is Even or Odd
Objective: Learn how to use the if-else statement in Python.
Task:
• Write a program that asks the user for a number.
• Check if the number is even or odd.
• Print the result.
Expected Output:
Enter a number: 7
7 is an odd number.

```python
def even(num): 
    if num % 2 == 0: 
        return 1;
    else:
        return 0;

# defining the 'even' function probably didn't help clean the code, but I wanted to remember
# how to use the syntax. 

num = int(input("Digit a number: "))

if even(num) == 1:
    print(f"{num} is an even number.\n")
else:
    print(f"{num} is an odd number.\n")
```

Exercise 17: Find the Largest Number
Objective: Learn how to compare values using if-elif-else.
Task:
• Take three numbers as input.
• Find and print the largest number.
Expected Output:
Enter three numbers: 5 12 9
The largest number is 12.

```python
def largest(n1,n2,n3):
    if n1 != n2 and n2 != n3:
        if n1 > n2 and n1 > n3:
            return n1;
        elif n2 > n3 and n2 > n3:
            return n2;
        elif n3 > n2 and n3 > n1:
            return n3;
    elif n1 == n2 and n2 != n3: 
        if n1 > n3:
            return n1;
        else:
            return n3;
    elif n1 == n3 and n1 != n2:
        if n1 > n2:
            return n1;
        else:
            return n2;
    elif n1 != n2 and n2 == n3:
        if n1 > n2:
            return n1;
        else:
            return n2;
    else:
        return n1;

# yeah, yeah. I know this doesn't look good. I was trying to punish myself and brute 
# force every combination possible to a return. 


n1 = int(input("first number: "))
n2 = int(input("second number: "))
n3 = int(input("third number: "))

print(f"The largest number is {largest(n1,n2,n3)}.")
```

Exercise 18: Check if a Year is a Leap Year
Objective: Use conditional logic to determine leap years.
Task:
• Ask the user for a year.
• Check if it is a leap year.
Expected Output:
Enter a year: 2024
2024 is a leap year.

```python
year = int(input("Enter a year: "))

if year % 4 == 0 and year % 100 != 0:
    print(f"{year} is a leap year.\n")
elif year % 100 == 0:
    if year % 400 == 0:
        print(f"{year} is a leap year.\n")
    else: 
        print(f"{year} is NOT a leap year.\n")
else: 
    print(f"{year} is NOT a leap year.\n")
```

Exercise 19: Print Numbers from 1 to N
Objective: Learn how to use a for loop.
Task:
• Take an integer input N.
• Print numbers from 1 to N.
Expected Output:
Enter a number: 5
1 2 3 4 5

```python
n = int(input("Enter N: "))

for i in range(1,n+1):
    print(i)
```
