Exercise 21: Sum of First N Natural Numbers
Objective: Learn how to use a loop to compute a sum.
Task:
• Take an integer input N.
• Compute the sum of numbers from 1 to N.
Expected Output:
Enter a number: 5
Sum of first 5 numbers: 15

```python
n = int(input("Enter a number: "))
soma = 0

for i in range(1,n+1):
    soma += i

print(f"Sum of first {n} numbers: {soma}")
    
```

Exercise 22: Factorial of a Number
Objective: Learn how to calculate factorial using loops.
Task:
• Take an integer input N.
• Compute N! (Factorial of N).
Expected Output:
Enter a number: 5
Factorial of 5 is 120.

```python
def factorial(n):
    prod = 1
    for i in range(1,n+1):
        prod = (prod * i)
    return prod;



n = int(input("Enter a number: "))
print(f"Factorial of {n} is {factorial(n)}")
```

Exercise 23: Reverse a Number
Objective: Use loops to reverse a number.
Task:
• Take an integer input.
• Reverse its digits and print the result.
Expected Output:
Enter a number: 1234
Reversed number: 4321

```python
num = int(input("Enter a number: "))
txt = str(num)    # txt becomes a vector of characters. 
rev = str()       # creating rev as an empty string. 

for i in range(len(txt),0,-1):       # in Python, vectors iniate in 0. 
    rev = rev + txt[i-1]            # the last 'len' element is not in the vector.
    
print(f"Reversed number: {rev}")
```

 Exercise 24: Multiplication Table
Objective: Use loops to generate a multiplication table.
Task:
• Take an integer input N.
• Print its multiplication table up to 10.
Expected Output:
Enter a number: 7
7 x 1 = 7
7 x 2 = 14
...
7 x 10 = 70

```python
n = int(input("Enter a number : "))

for i in range(1,10):
    print(f"{n} x {i} = {n*i}")    
```

Exercise 25: Count Digits in a Number
Objective: Use loops to count the digits in a number.
Task:
• Take an integer input.
• Count and print the number of digits.
Expected Output:
Enter a number: 56473
The number has 5 digits.

```python
num = int(input("Enter a number: "))
count = 0

while num > 0:
    num = int(num / 10)      # int division. 
    count += 1

print(f"The number has {count} digits")
```
