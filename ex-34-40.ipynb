Exercise 35: Sort a List of Student Scores
Objective: Learn how to sort a list.
Task:
• Given the list of scores [88, 92, 75, 100, 89], sort them in ascending order.
Expected Output:
[75, 88, 89, 92, 100]

```python
scores = [88,92,75,100,89]

scores.sort()   

print(f'Updated Scores: {scores}')

scores_ascending = sorted(scores)                # 'sorted' instead of 'sort()' to create a new variable.
scores_descending = sorted(scores,reverse=True)    # reverse does the opposite. 

scores.sort()

print(f"Ascending Scores: {scores_ascending}\nDescending Scores: {scores_descending}\n")
```

Exercise 38: Store Student Grades in a Dictionary
Objective: Store and retrieve student grades using a dictionary.
Task:
• Create a dictionary where student names are keys and their grades are values.
• Retrieve the grade of ”Sarah”.
Expected Output:

```python
grades = {'Gabriel': 'A+','Sarah':'A','John':'C','Marcus':'F'}

#grades.keys()
#grades.values()

name = str(input("Write down a student's name: "))

if(name in grades) == False:
    print('Error: Name not found.\n')
else:
    print(f"{name}'s Grade: {grades.get(name)}\n")
```

Exercise 40: Identify Unique Words in a Text
Objective: Use a set to find unique words in a text.
Task:
• Extract all unique words from the text: ”Python is amazing and Python is powerful”.
Expected Output:
{'Python', 'is', 'amazing', 'and', 'powerful'}

```python
txt = "Python is amazing and Python is powerful."

split = txt.split()        # split() creates a list 

split_set = set(split)     # turning a list into a set. 

print(sorted(split_set))    # 'sorted' to put it in alphabetic order.
```
